```python
import os
import numpy as np
import matplotlib.pyplot as plt
import random

# Ruta base donde están los pacientes
path_slices = r"data\LIDC-IDRI_images_npy_3d_Slices"

# Seleccionar aleatoriamente un paciente
pacientes = [p for p in os.listdir(path_slices) if os.path.isdir(os.path.join(path_slices, p))]
paciente = random.choice(pacientes)
print(f"Paciente seleccionado: {paciente}")

# Ruta del paciente
path_paciente = os.path.join(path_slices, paciente)

# Listar todos los nódulos .npy
nodulos = [n for n in os.listdir(path_paciente) if n.endswith('.npy')]
print(f"Nódulos encontrados: {len(nodulos)}")

# Calcular tamaño de la cuadrícula
cols = 4
rows = int(np.ceil(len(nodulos) / cols))

# Crear figura
plt.figure(figsize=(cols * 3, rows * 3))

# Plotear cada nódulo
for i, nodule in enumerate(nodulos):
    nodule_path = os.path.join(path_paciente, nodule)
    nodule_data = np.load(nodule_path)
    
    plt.subplot(rows, cols, i + 1)
    plt.imshow(nodule_data, cmap='gray')
    plt.title(f"Nódulo {i+1}")
    plt.axis('off')

plt.suptitle(f"Nódulos del paciente {paciente}", fontsize=16)
plt.tight_layout()
plt.show()
```
